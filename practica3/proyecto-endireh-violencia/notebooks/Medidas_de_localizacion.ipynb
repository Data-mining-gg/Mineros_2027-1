# Medidas de localización

```python
import numpy as np
import polars as pl
import sys
from pathlib import Path

#Para poder utilizar archivos del proyecto
ROOT = Path.cwd().parent 
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.cleaning.preprocessing import pipeline_preprocesamiento
df = pipeline_preprocesamiento()
df = df.filter(
    (pl.col("edad_primer_union").is_not_null()) & 
    (pl.col("edad_primer_union") < 90) &         
    
    (pl.col("num_hijos").is_not_null()) & 
    (pl.col("num_hijos") < 90)       
)
```

## Cálculo de medidas de tendencia central

```python
medidas = df.select([
    pl.col("edad_primer_union").mean().alias("media_edad_primer_union"),
    pl.col("edad_primer_union").median().alias("mediana_edad_primer_union"),
    pl.col("edad_primer_union").mode().first().alias("moda_edad_primer_union"),
    pl.col("edad_primer_union").quantile(0.10).alias("P10_edad_primer_union"),
    pl.col("edad_primer_union").quantile(0.25).alias("Q1_edad_primer_union"),
    pl.col("edad_primer_union").quantile(0.50).alias("Q2_edad_primer_union"),
    pl.col("edad_primer_union").quantile(0.75).alias("Q3_edad_primer_union"),
    pl.col("edad_primer_union").quantile(0.90).alias("P90_edad_primer_union"),
    
    pl.col("num_hijos").mean().alias("media_hijos"),
    pl.col("num_hijos").median().alias("mediana_hijos"),
    pl.col("num_hijos").mode().first().alias("moda_hijos"),
    pl.col("num_hijos").quantile(0.10).alias("P10_hijos"),
    pl.col("num_hijos").quantile(0.25).alias("Q1_hijos"),
    pl.col("num_hijos").quantile(0.50).alias("Q2_hijos"),
    pl.col("num_hijos").quantile(0.75).alias("Q3_hijos"),
    pl.col("num_hijos").quantile(0.90).alias("P90_hijos")


])
pl.Config.set_tbl_rows(-1)
medidas_vertical = medidas.transpose(include_header=True, header_name="Metrica")
medidas_vertical
```

## Calculo de media ponderada

```python

edad = df["edad_primer_union"].to_numpy()
hijos = df["num_hijos"].to_numpy()
pesos = df["factor_expansion"].to_numpy()

media_edad_ponderada = np.average(edad, weights=pesos)
media_hijos_ponderada = np.average(hijos, weights=pesos)

print(f"La media ponderada de la edad de la primera union es: {media_edad_ponderada}")
print(f"La media ponderada de el número de hijos es: {media_hijos_ponderada}")
```

## Relación entre media ponderada y media

Podemos ver que la media ponderada de la edad de la primera union coincide con la media con una diferencia de 0.3. La media ponderada del número de hijos tambien coincide, sin embargo hay una diferencia de aproximadamente 0.1, lo 
cuál esta bien. Lo que significa que nuestra muestra es relativamente cercana la población en general y no
presenta desvios muy grandes.
